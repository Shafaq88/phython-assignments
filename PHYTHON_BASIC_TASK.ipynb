```python
#task 1 CINEMA TICKET PRICE

name = input("enter your name")
age = int(input("enter your age"))

if age < 5 or age > 100:
    ticket_price = 0
elif age >= 5 and age <= 12:
    ticket_price = 300
elif age >= 13 and age <= 59:
    ticket_price = 700
elif age >= 60 and age <= 100:
    ticket_price = 400

print("Customer:", name)
print("Ticket Price: Rs.", ticket_price)
```

```python
#task 2 TEMPRATURE CONVERTOR

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32

if celsius < 10:
    weather = "Very Cold"
elif celsius <= 19:
    weather = "Cold"
elif celsius <= 29:
    weather = "Normal"
else:
    weather = "Hot"

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
print("Weather:", weather)
```

```python
#task 3 MOBILE RECHARGE CALCULATOR

mobile = input("Enter mobile number: ")
amount = float(input("Enter recharge amount: "))

tax = amount * 3 / 100
final_balance = amount - tax

print("Mobile Number:", mobile)
print("Recharge Amount:", amount)
print("Tax:", tax)
print("Final Balance:", final_balance)
```

```python
#task 5 SECURE PIN CHECKER

secret_pin = "1234"
attempts = 0

while attempts < 3:
    pin = input("Enter PIN: ")
    attempts = attempts + 1

    if pin == secret_pin:
        print("Access Granted")
        break
    else:
        remaining = 3 - attempts
        print("Incorrect PIN")
        print("Remaining attempts:", remaining)

if attempts == 3 and pin != secret_pin:
    print("Account Blocked")
```

```python
#task 4 WEEKLY STEPS TRACKER

total_steps = 0

for day in range(7):
    steps = int(input("Enter steps: "))
    total_steps = total_steps + steps

average = total_steps / 7

if average >= 5000:
    status = "Active Week"
else:
    status = "Needs More Activity"

print("Total Steps:", total_steps)
print("Daily Average:", average)
print("Status:", status)
```
